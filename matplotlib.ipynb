```python
import matplotlib.pyplot as plt
import numpy as np
x=np.array([2022,2023,2024,2025])
y1=np.array([2,3,4,5])
y2=np.array([4,5,6,7])
y3=np.array([12,21,22,20])
y4=np.array([23,25,26,27])


line_style=dict(marker='o',
                markersize=8,
                markerfacecolor='Red',
                markeredgecolor='Red',
                linestyle='solid',
                linewidth=2,
                color='Blue')
plt.plot(x,y1, **line_style)
plt.plot(x,y2, **line_style)
plt.plot(x,y3, **line_style)
plt.plot(x,y4, **line_style)

plt.show()
```

# Grid

###

 grid() = grid() function in Matplotlib is used to enhance the readability and interpretability 
 of plots by adding a grid of horizontal and/or vertical lines. These lines serve as visual aids, 
 making it easier to analyze and understand the data presented in a graph.


```python
x=np.array([1,2,3,4,5])
y=np.array([5,10,15,20,25])

plt.grid(linewidth=2,
         color='Yellow',
         linestyle='-')
plt.plot(x,y, marker='v')
plt.show()
```

```python
x=np.array([1,2,3,4,5])
y=np.array([5,10,15,20,25])

plt.grid(linewidth=2,
         color='Yellow',
         linestyle=':')
plt.plot(x,y, marker='o')
plt.show()
```

## Bar Chart

```python
fruits =np.array(['Apples','Bananas','cherries','Dates'])
sales =np.array([500,350,300,450])

line_style = dict(fontsize=15,
                  family='Arial',
                  fontweight='bold',
                  color='Blue')

plt.bar(fruits,sales, color="Red")
plt.title('Fruit Sales',fontsize=15,color='skyblue')
plt.xlabel('Fruits',**line_style)
plt.ylabel('Sales',**line_style)
plt.show()                
                                
                
```

## Pie Chart

```python
x=np.array([25,50,5,20,8])
mylabels=["Apples","Bananas","Cherries",'Dates','Oranges']
colors = ['red', 'blue', 'green', 'orange',"red"]

plt.pie(x,labels=mylabels,
        autopct="%1.1f%%",
        colors=colors,)

plt.title("Fruit Sales")
plt.show()
```

## Scatter plot

```python
x1=np.array([0,1,5,7,4,4,3,8])#stdent study hour
y1=np.array([10,20,50,65,75,60,55,85])#students marks
plt.title("Test Scores")
plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.scatter(x1,y1)
plt.show()
```

## Histogram

```python
x=np.random.normal(loc=40, scale=10,size=100)
plt.hist(x)
plt.show
```

## Sub plot

```python
import matplotlib.pyplot as plt
import numpy as np

x=np.array([1,2,3,4,5])
figure,axes=plt.subplots(2,2)

axes[0,0].plot(x,x*2,color='Red')
axes[0,0].set_title('Multiple x')

axes[0,1].plot(x,x**2,color='Yellow')
axes[0,1].set_title('x**2')

axes[1,0].plot(x,x**3,color='Blue')
axes[1,0].set_title('x**3')

axes[1,1].plot(x,x**4,color='Orange')
axes[1,1].set_title('x**4')

plt.tight_layout()
plt.show()






```
