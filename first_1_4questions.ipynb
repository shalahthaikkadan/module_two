### 1. Series from a List with Filtering

Task:

Create a Series from [10, 20, 30, 40, 50].

Use loc[] to access the element with index 3.

Use iloc[] to access the second element.

Filter elements greater than 25.

```python
import pandas as pd

s = pd.Series([10, 20, 30, 40, 50])

print(s.loc[3])  
print()     
print(s.iloc[1])  
print()    

print(s[s > 25])      
```

### 2. Series from a Dictionary

Task:

Create a Series from {"A": 5, "B": 15, "C": 25, "D": 35}.

Access the value of key "B" using loc[].

Access the first element using iloc[].

Filter values greater than 20.

```python
s = pd.Series({"A": 5, "B": 15, "C": 25, "D": 35})

print(s.loc["B"]) 
print()    
print(s.iloc[0])
print()  

print(s[s > 20])      
```

4. Conditional Filtering in a DataFrame

Task:

Filter rows where Age > 28.

Filter rows where City == "Delhi".

Print both filtered DataFrames.



```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, 30, 35],
    "City": ["Delhi", "Mumbai", "Chennai"]
})
print(df)
print()
print(df.loc[1])
print()
print(df.iloc[0])
print()
print(df[df["Age"] > 28])
print()
print(df[df["City"] == "Delhi"])
```
