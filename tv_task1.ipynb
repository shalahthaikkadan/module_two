```python
import pandas as pd

df = pd.read_csv('TV_Final.csv')
df
```

### 1. Filtering Based on Condition

Task:

Display all rows where the Brand is "SAMSUNG" AND the Selling Price is less than 50000.

```python
print(df[(df["Brand"] == "SAMSUNG") & (df["Selling Price"] < 50000)])
```

### 2. Conditional Filtering with Multiple Conditions

Task:

Display only the Brand, Size, and Rating columns for HD LED products where the Operating System is "Android " AND Rating is greater than or equal to 4.0.

```python
print(df.loc[
    (df["Resolution"] == "HD LED") & (df["Rating"] >= 4.0),
    ["Brand", "Size ", "Rating"]
])
```

### 3. Drop Rows with Missing Values

Task:

Drop all rows where the Rating column has NaN values.

```python

df = df.dropna(subset=["Rating"])
print(df)
```

### 4. Fill Missing Values

Task:

Replace all NaN values in the Operating System column with the string "Unknown".

```python
df["Operating System"] = df["Operating System"].fillna("Unknown")
print(df["Operating System"])
```

### 5. Rename Columns

Task:

Rename the column "Selling Price" to "Current Price" and "Original Price" to "MRP".

```python
df = df.rename(columns={
    "Selling Price": "Current Price",
    "Original Price": "MRP"
})

print(df.columns)
print(df)
```

### 6. Count Occurrences Using value_counts()

Task:

Find how many HD LED products are available for each Brand using value_counts().

```python
print(df[df["Resolution"] == "HD LED"]["Brand"].value_counts())
```

### 7. Grouping and Aggregation

Task:

Group the DataFrame by Brand and calculate the sum of Selling Price for each brand.

Then, sort the result in descending order of average selling price.

```python
result = df.groupby("Brand")["Current Price"].agg(
    Total_Price="sum",
    Average_Price="mean"
)

print(result.sort_values("Average_Price", ascending=False))
```

### 8. Sorting Values

Task:

Sort the DataFrame first by Selling Price in ascending order and then by Rating in descending order.

```python
print(df.sort_values(
    by=["Current Price", "Rating"],
    ascending=[True, False]
))
```

### 9. GroupBy with Sort

Task:

Find the top 3 brands with the highest selling price.

Display their brand names.

```python
top3 = df.groupby("Brand")["Current Price"].sum().nlargest(3)

print(top3)
```
