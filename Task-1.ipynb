### 1. Create a DataFrame
- Create a DataFrame with columns `Name`, `Age`, and `Marks`.
- Include 4 student records with at least one missing value.
- Hint: Use a dictionary and `pd.DataFrame()`.

```python
import pandas as pd

data = {
    'Name': ['Anu', 'Rahul', 'Manu', 'Neha'],
    'Age': [20, None, 24, 22],
    'Marks': [85, 78, 92, 88]
}

df = pd.DataFrame(data)
print(df)
```

### 2. Add a New Column
- Add a new column `Grade` with letter grades (A, B, C, etc.).

```python
df['Grade'] = ['A+', 'A', 'B+', 'B']
print(df)
```

### 3. Access Specific Columns
- Display only the `Name` and `Marks` columns.

```python
print(df[['Name', 'Marks']])
```

### 4. Count Missing Values
- Count the number of missing values in each column.
- Hint: `df.isnull().sum()`.

```python
print(df.isnull().sum())
```

### 5. Handle Missing Values
- Replace all missing values in `Age` with the mean age.
- Hint: `df['Age'].fillna(df['Age'].mean(), inplace=True)`.

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
print(df)
```

### 6. Modify a Specific Cell
- Change the `Name` in the 3rd row (index 2) to `Sonu`.
- Hint: `df.loc[2, 'Name'] = 'Sonu'`.

```python
df.loc[2, 'Name'] = 'Sonu'
print(df)
```

### 7. Add New Rows to the DataFrame
- Add two new rows:
  - `{'Name': 'Kumar', 'Age': 20, 'Marks': 90}`
  - `{'Name': 'Rahul', 'Age': 26, 'Marks': 85}`
- Hint: Use `pd.concat()`.

```python
new_names = pd.DataFrame([
    {'Name': 'Kumar', 'Age': 20, 'Marks': 90},
    {'Name': 'Rahul', 'Age': 26, 'Marks': 85}
])

df = pd.concat([df, new_names], ignore_index=True)
print(df)
```

### 11. Aggregate Functions
Calculate and display:
- Mean Age
- Minimum Age
- Maximum Age
- Count of Age values

```python
print("Mean Age:", df['Age'].mean())
print("Minimum Age:", df['Age'].min())
print("Maximum Age:", df['Age'].max())
print("Count of Age values:", df['Age'].count())
```

### 12. Grouping and Aggregation
- Group the DataFrame by `Name` and compute the mean `Age` for each name.
- Hint: `df.groupby('Name')['Age'].mean()`.

```python
df.groupby('Name')['Age'].mean()
```
