```python
import pandas as pd
df = pd.read_csv("btc-usd-max.csv")
print(df)
```

```python
print(df.head())
```

```python
print(df.info())
```

```python
print(df.isnull().sum())
```

```python
df.drop_duplicates(inplace=True)
df
```

```python
df["snapped_at"] = pd.to_datetime(df["snapped_at"], utc=True)
df
```

```python
columns = ["price", "market_cap", "total_volume"]

for col in columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
df
```

```python
df.dropna(inplace=True)
```

```python
df = df[
    (df["price"] >= 0) &
    (df["market_cap"] >= 0) &
    (df["total_volume"] >= 0)
]
```

```python
df.sort_values(by="snapped_at", inplace=True)
```

```python
df.reset_index(drop=True, inplace=True)
```

```python
print("\nCleaned Dataset:")
print(df.head())
```

```python
print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nFinal Dataset Shape:", df.shape)
```

```python
df.to_csv("btc_usd_cleaned.csv", index=False)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model

df=pd.read_csv('sales_data.csv')
df["Unit_Price"] = df["Unit_Price"].fillna(df["Unit_Price"].median())
df["Total_Sales"] = df["Total_Sales"].fillna(df["Total_Sales"].median())
plt.xlabel("Total_Sales")
plt.ylabel("Unit_Price")
plt.scatter(df.Total_Sales,df.Unit_Price)
plt.show()


reg=linear_model.LinearRegression()
reg.fit(df[['Total_Sales']],df.Unit_Price)


res=reg.predict([[21000]])
print(res)
print(reg.coef_)
print(reg.intercept_)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model


df = pd.read_csv('btc_usd_cleaned.csv')


plt.xlabel("Market Cap")
plt.ylabel("Bitcoin Price")

plt.scatter(df.market_cap, df.price)
plt.show()

reg = linear_model.LinearRegression()

reg.fit(df[['market_cap']], df.price)


res = reg.predict([[21000]])

print("Predicted Price:", res)
print("Coefficient:", reg.coef_)
print("Intercept:", reg.intercept_)
```
