1.Create a line graph to show the population growth of India from 2000 to 2020.

years = [2000, 2005, 2010, 2015, 2020]

population = [1.0, 1.1, 1.2, 1.3, 1.38]

- Add labels for x and y axes.
- Add title "Population Growth of India (2000–2020)".
- Display grid lines.

```python

import matplotlib.pyplot as plt

years = [2000, 2005, 2010, 2015, 2020]
population = [1.0, 1.1, 1.2, 1.3, 1.38]

plt.plot(years, population, marker='o')

plt.xlabel("Year")
plt.ylabel("Population")
plt.title("Population Growth of India (2000–2020)")

plt.grid(
    linewidth=2,
    color='yellow',
    linestyle='-'
)

plt.show()
```

5. Draw a pie chart showing the market share of mobile brands.

brands = ['Samsung', 'Apple', 'Xiaomi', 'OnePlus', 'Others']

share = [30, 25, 20, 15, 10]

- Display percentage values.
- Explode the slice with the highest share.

```python

import numpy as np

brands = ['Samsung', 'Apple', 'Xiaomi', 'OnePlus', 'Others']
share = np.array([30, 25, 20, 15, 10])

explode = [0.1, 0, 0, 0, 0]

colors = ['blue', 'red', 'orange', 'green', 'purple']

plt.pie(share, labels=brands,
        autopct="%1.1f%%",
        colors=colors,
        explode=explode)

plt.title("Mobile Brands Market Share")
plt.show()
```
