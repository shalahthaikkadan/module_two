# Sample Dataset 5️⃣: Category and Quantity

df_sales = pd.DataFrame({
    'Product Category': ['Electronics', 'Clothing', 'Groceries', 'Toys', 'Books'],
    'Quantity Sold': [120, 200, 150, 90, 130]
})

### 5️⃣ Create a subplot with two plots side by side:
- Left plot: Bar chart of 'Product Category' vs 'Quantity Sold'.
- Right plot: Pie chart showing 'Category' wise sales percentage.
```

````python
import pandas as pd
import matplotlib.pyplot as plt

df_sales = pd.DataFrame({
    'Product Category': ['Electronics', 'Clothing', 'Groceries', 'Toys', 'Books'],
    'Quantity Sold': [120, 200, 150, 90, 130]
})

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.bar(df_sales['Product Category'], df_sales['Quantity Sold'])
plt.xlabel("Product Category")
plt.ylabel("Quantity Sold")
plt.title("Quantity Sold by Category")
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
plt.pie(df_sales['Quantity Sold'], labels=df_sales['Product Category'], autopct='%1.1f%%')
plt.title("Category-wise Sales Percentage")

plt.tight_layout()
plt.show()
````

````python
import pandas as pd
import matplotlib.pyplot as plt

df_company = pd.DataFrame({
    'Year': [2018, 2019, 2020, 2021, 2022, 2023],
    'Revenue': [50000, 60000, 70000, 85000, 95000, 110000],
    'Ad Spend': [5000, 7000, 9000, 10000, 11000, 13000]
})

figure, axes = plt.subplots(2, 1)

axes[0].plot(df_company['Year'], df_company['Revenue'], color='Blue')
axes[0].set_title('Year vs Revenue')

axes[1].scatter(df_company['Ad Spend'], df_company['Revenue'], color='Red')
axes[1].set_title('Advertising Spend vs Revenue')

plt.tight_layout()
plt.show()
````
